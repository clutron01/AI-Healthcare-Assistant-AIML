```python
from transformers import AutoTokenizer
from transformers import AutoModelForCausalLM
from transformers import pipeline
```

```python
model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
model = AutoModelForCausalLM.from_pretrained(model_name)
```

```python
chatbot = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)
```

```python
prompt = """
You are an experienced medical assistant.

Answer the following medical question in a clear and simple way.

Question:
What are the symptoms of diabetes?
"""
```

```python
response = chatbot(
    prompt,
    max_new_tokens=200,
    do_sample=True,
    temperature=0.7
)
```

```python
print(response[0]["generated_text"])
```

```python
prompt = """
You are an experienced medical doctor.

Context:
The user is a patient with no medical background who wants to understand diabetes.

Task:
Explain diabetes in a simple and accurate way.

Constraints:
- Use simple English.
- Do not use difficult medical terms.
- Do not diagnose the patient.
- Do not recommend medicines.
- Keep the explanation under 300 words.

Output Format (Markdown):

# Disease Overview

# Symptoms

# Risk Factors

# Prevention

# When to Consult a Doctor
"""
```

```python
response = chatbot(
    prompt,
    max_new_tokens=300,
    do_sample=True,
    temperature=0.7
)

print(response[0]["generated_text"])
```

```python
role = input("Enter your role (Doctor/Nurse/Medical Student/Patient): ")

question = input("Enter your medical question: ")

prompt = f"""
You are a healthcare AI assistant.

The user's role is: {role}

Instructions:

If the user is a Doctor:
- Give detailed clinical information.
- Use professional medical terminology.

If the user is a Nurse:
- Focus on patient care, monitoring and nursing responsibilities.

If the user is a Medical Student:
- Explain concepts step-by-step for learning.
- Include simple explanations.

If the user is a Patient:
- Use simple language.
- Avoid medical jargon.
- Do not diagnose.
- Recommend consulting a healthcare professional when necessary.

Question:
{question}

Provide a clear and well-structured response in Markdown.
"""
```

```python
response = chatbot(
    prompt,
    max_new_tokens=300,
    do_sample=False
)

print(response[0]["generated_text"])
```
